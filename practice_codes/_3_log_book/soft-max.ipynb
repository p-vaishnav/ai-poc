```python
import math
import numpy as np
```

```python
a = np.array([0.25, 0.12, 0.21, 0.36, -0.9])
```

```python
type(a)
```

```python
a_new = list()
for element in a:
    a_new.append(np.exp(element))
```

```python
a_new
```

```python
total = sum(a_new)
```

```python
total
```

```python
a_softmax = list()
for element in a_new:
    a_softmax.append(element / total)
```

```python
a_softmax
```
