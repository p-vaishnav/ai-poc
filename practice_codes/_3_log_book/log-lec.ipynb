```python
# odds and logarithmic practice
import numpy as np

prior_odds = 0.4 / 0.6
LR_F = (3 / 4) / (1 / 6) # quick reply > 4.5
LR_Q = 0.9 / 0.3 # reverse question: 3

# odds form: total
odds = prior_odds * LR_F * LR_Q
print("odds form :", round(odds, 3))

# log form: total
points = np.log(prior_odds) + np.log(LR_F) + np.log(LR_Q)

print("log form :", round(points, 3))
print("ln(9)    :", round(np.log(9), 3))
print("e^points :", round(np.exp(points), 3))
```

```python
import numpy as np
import matplotlib.pyplot as plt

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

z = np.linspace(-6, 6, 300)


journey = {"Prior": np.log(2/3),
           "After Fast reply": np.log(3),
           "After Question": np.log(9)}

plt.plot(z, sigmoid(z), label="sigmoid(z)")
for name, zi in journey.items():
    plt.scatter(zi, sigmoid(zi), s=80, zorder=3)
    plt.annotate(f"{name}\nz={zi:.1f}, p={sigmoid(zi):.2f}",
                 (zi, sigmoid(zi)), textcoords="offset points", xytext=(10, -25))
plt.axhline(0.5, linestyle="--", color="gray")
plt.xlabel("z (log odds points)")
plt.ylabel("P(Interested)")
plt.title("Belief journey on the sigmoid")
plt.grid(alpha=0.3)
plt.legend()
plt.savefig("sigmoid_journey.png", dpi=120, bbox_inches="tight")
plt.show()
```
