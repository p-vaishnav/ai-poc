# Computational Graph: Partial Derivatives of $L$

This notebook calculates the partial derivative of $L$ with respect to every variable in the computational graph, using the **chain rule**.

We use the definitions from the handwritten notes:

$$
a=5,\quad b=3,\quad c=-2,\quad g=6
$$
$$
e=\frac{a}{b},\qquad f=c+e,\qquad L=f\,g
$$

We'll first perform the forward pass, then derive each gradient, and finally verify the results with Python.

## 1. Forward pass

Calculate each intermediate variable in order: `e`, `f`, and `L`.

```python
a = 5
b = 3
c = -2
g = 6

e = a / b
f = c + e
L = f * g

print(f"a = {a}")
print(f"b = {b}")
print(f"c = {c}")
print(f"e = a / b = {e}")
print(f"f = c + e = {f}")
print(f"g = {g}")
print(f"L = f * g = {L}")
```

## 2. Partial derivative of $L$ with respect to $f$

Since $L=fg$, treat $g$ as constant:

$$
\boxed{\frac{\partial L}{\partial f}=g}
$$

```python
dL_df = g
print("∂L/∂f =", dL_df)
```

## 3. Partial derivative of $L$ with respect to $g$

Treat $f$ as constant:

$$
\boxed{\frac{\partial L}{\partial g}=f}
$$

```python
dL_dg = f
print("∂L/∂g =", dL_dg)
```

## 4. Partial derivative of $L$ with respect to $c$

Because $f=c+e$, we have $\partial f/\partial c=1$. By the chain rule:

$$
\frac{\partial L}{\partial c}
=\frac{\partial L}{\partial f}\frac{\partial f}{\partial c}
=g\cdot 1
$$

$$
\boxed{\frac{\partial L}{\partial c}=g}
$$

```python
df_dc = 1
dL_dc = dL_df * df_dc
print("∂L/∂c =", dL_dc)
```

## 5. Partial derivative of $L$ with respect to $e$

Since $f=c+e$, $\partial f/\partial e=1$:

$$
\frac{\partial L}{\partial e}
=\frac{\partial L}{\partial f}\frac{\partial f}{\partial e}
=g\cdot 1
$$

$$
\boxed{\frac{\partial L}{\partial e}=g}
$$

```python
df_de = 1
dL_de = dL_df * df_de
print("∂L/∂e =", dL_de)
```

## 6. Partial derivative of $L$ with respect to $a$

Because $e=a/b$:

$$
\frac{\partial e}{\partial a}=\frac{1}{b}
$$

Apply the chain rule along $a\to e\to f\to L$:

$$
\frac{\partial L}{\partial a}
=\frac{\partial L}{\partial f}
\frac{\partial f}{\partial e}
\frac{\partial e}{\partial a}
=g\cdot 1\cdot\frac{1}{b}
$$

$$
\boxed{\frac{\partial L}{\partial a}=\frac{g}{b}}
$$

```python
de_da = 1 / b
dL_da = dL_df * df_de * de_da
print("∂L/∂a =", dL_da)
```

## 7. Partial derivative of $L$ with respect to $b$

For $e=a/b$:

$$
\frac{\partial e}{\partial b}=-\frac{a}{b^2}
$$

The negative sign is important. Apply the chain rule:

$$
\frac{\partial L}{\partial b}
=\frac{\partial L}{\partial f}
\frac{\partial f}{\partial e}
\frac{\partial e}{\partial b}
=g\cdot 1\cdot\left(-\frac{a}{b^2}\right)
$$

$$
\boxed{\frac{\partial L}{\partial b}=-\frac{ga}{b^2}}
$$

```python
de_db = -a / (b ** 2)
dL_db = dL_df * df_de * de_db
print("∂L/∂b =", dL_db)
```

## 8. Summary of all partial derivatives

| Variable | Derivative | Numerical value |
|---|---|---:|
| $f$ | $\partial L/\partial f=g$ | $6$ |
| $g$ | $\partial L/\partial g=f$ | $-1/3$ |
| $c$ | $\partial L/\partial c=g$ | $6$ |
| $e$ | $\partial L/\partial e=g$ | $6$ |
| $a$ | $\partial L/\partial a=g/b$ | $2$ |
| $b$ | $\partial L/\partial b=-ga/b^2$ | $-10/3$ |

Note: these are partial derivatives of the expression $L=fg$, treating the other input at each local operation as independent. The derivative with respect to $g$ is $f$, even though the chosen numerical value of $g$ is 6.

## 9. Verify the results with assertions

This cell checks the computed gradients against their expected values.

```python
assert abs(dL_df - 6) < 1e-12
assert abs(dL_dg - (-1/3)) < 1e-12
assert abs(dL_dc - 6) < 1e-12
assert abs(dL_de - 6) < 1e-12
assert abs(dL_da - 2) < 1e-12
assert abs(dL_db - (-10/3)) < 1e-12

print("All gradient checks passed.")
```
