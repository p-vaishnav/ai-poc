# Loss = a * b + c  (one gradient-descent step)

Given: `a = 3`, `b = -2`, `c = 5`, learning rate `lr = 0.1`  (c stays fixed)

```python
a, b, c = 3, -2, 5
lr = 0.1
```

## a) Current loss

Just plug the numbers in: 3 * (-2) + 5 = -6 + 5 = **-1**

```python
loss = a * b + c
print("loss =", loss)
```

## b) Derivatives

- `dL/da`: treat `b` and `c` as fixed numbers. Only `a*b` has `a` in it, and its slope is `b`. So `dL/da = b = -2`.
- `dL/db`: same idea, the slope is `a`. So `dL/db = a = 3`.

```python
dL_da = b
dL_db = a
print("dL/da =", dL_da)
print("dL/db =", dL_db)
```

## c) One gradient-descent step

Rule: `new = old - lr * derivative`

- `a`: 3 - 0.1 * (-2) = **3.2**
- `b`: -2 - 0.1 * (3) = **-2.3**

```python
new_a = a - lr * dL_da
new_b = b - lr * dL_db
print("new a =", new_a)
print("new b =", new_b)
```

## d) Recompute the loss

3.2 * (-2.3) + 5 = -7.36 + 5 = **-2.36**

It went from -1 to -2.36, so the loss went **down**.

```python
new_loss = new_a * new_b + c
print("old loss =", loss)
print("new loss =", round(new_loss, 4))
print("went down?", new_loss < loss)
```

## e) Why the negative of the derivative?

**The sign tells you which way is uphill.**
- `dL/da = -2` (negative): raising `a` makes the loss go down. So `a` went up (3 -> 3.2).
- `dL/db = +3` (positive): raising `b` makes the loss go up. So `b` went down (-2 -> -2.3).

**Why subtract it:** the derivative always points toward *higher* loss. We want *lower* loss, so we step the opposite way. Like standing on a hill in fog: the slope tells you which way is up, so you walk the other way to go down.

**The size sets the step:** `b` has the bigger derivative (3 vs 2), so it moved more (0.3 vs 0.2).

**One line:** the derivative points toward increasing loss, so we subtract it to make the loss go down.

## f) "Unbounded below"

It means the loss has **no bottom**: it can keep getting smaller forever.

If `a` is a huge positive number and `b` is a huge negative number, `a*b` is hugely negative. Make them bigger and the loss gets even smaller, with no limit.

**For part (d):** the loss did go down (-1 -> -2.36), but that doesn't mean we're getting closer to a best answer. If we keep stepping, it just keeps dropping, faster each time.

**Why "reduce the loss" is a bad goal here:** it only makes sense when there is a lowest point to reach, like walking down into a valley and stopping at the bottom. Here there is no valley, just a cliff that never ends.

Let's run 10 more steps and watch it:

```python
a, b = 3.0, -2.0
for step in range(11):
    print(f"step {step:2d}: a={a:8.3f}  b={b:8.3f}  loss={a * b + c:9.3f}")
    grad_a, grad_b = b, a
    a, b = a - lr * grad_a, b - lr * grad_b
```

The loss never settles. `a` keeps growing, `b` keeps getting more negative, and the loss keeps dropping faster. There is no "best" answer to land on.

**One line:** the loss has no lowest value, so gradient descent never finishes. It just pushes the loss lower forever.
