```python
import numpy as np
import matplotlib.pyplot as plt

z = -1 / 5
xl = np.linspace(-2, -0.001, 200)
xr = np.linspace(0.001, 2, 200)

plt.figure(figsize=(6, 4.5))
plt.plot(xl, -2*xl  + 5*z + 4, 'b-', label='x < 0: -2x + 5z + 4')
plt.plot(xr, 2*xr  + 3, 'g-', label='x > 0: -2x + 3')

plt.plot(0, 3, 'ko', mfc='white', ms=9, label='both limits = 3')
plt.plot(0, 5*z, 'ro', ms=8, label='both limits = 3')

plt.axvline(0, color='gray', lw=0.5)
plt.grid(alpha=0.9)

plt.xlabel('x'); plt.ylabel('f(x)')
plt.title('z = -1/5: limits agree (3) but f(0) = -1')
plt.show()
```
